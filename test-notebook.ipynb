```python
import os
from openai import OpenAI, OpenAIError, AuthenticationError


def test_openai_key():
    key = os.getenv("OPENAI_API_KEY")
    if not key:
        print("❌ Environment variable OPENAI_API_KEY is not set.")
        return False

    # Instantiate the new v1 client
    client = OpenAI(api_key=key)

    try:
        # A minimal call to verify your key (you can pick any model you have access to)
        client.chat.completions.create(
            model="gpt-3.5-turbo",
            messages=[{"role": "user", "content": "ping"}],
            max_tokens=1,
        )
        print("✅ Your OpenAI API key is working!")
        return True

    except AuthenticationError:
        print("❌ AuthenticationError: Your API key is invalid or has expired.")
        return False

    except OpenAIError as e:
        # Catches other API errors (rate limits, invalid model, etc.)
        print(f"⚠️ OpenAIError: {e}")
        return False

if __name__ == "__main__":
    test_openai_key()
```
